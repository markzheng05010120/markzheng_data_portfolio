```python
import pandas as pd
```

```python
df = pd.read_csv('cleaned_data.csv')
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0, 30, 50, 100],
    labels=['0-30', '30-50', '50+'])
```

```python
df['glucose_group'] = pd.qcut(
    df['avg_glucose_level'],
    q=3,
    labels=['low', 'medium', 'high'])
```

```python
dashboard_df = df[[
    'age',
    'age_group',
    'stroke',
    'avg_glucose_level',
    'glucose_group',
    'bmi']].copy()
```
