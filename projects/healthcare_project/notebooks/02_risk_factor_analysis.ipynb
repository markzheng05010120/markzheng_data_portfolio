```python
import pandas as pd
```

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
df = pd.read_csv('cleaned_data.csv')
```

```python
df.head()
```

```python
df.describe()
# basic statistics for numerical columns
```

```python
df['stroke'].value_counts()
```

```python
df['stroke'].value_counts(normalize=True)
```

```python
df.groupby('stroke')['age'].mean()
# compare average age between stroke and non-stroke patients
```

```python
sns.boxplot(x='stroke', y='age', data=df)
plt.title('Age vs Stroke')
plt.show()
```

```python
pd.crosstab(df['hypertension'], df['stroke'])
```

```python
pd.crosstab(df['hypertension'], df['stroke'], normalize='index')
```

```python
# analyze stroke rate for patients with hypertension
```

```python
sns.boxplot(x='stroke', y='bmi', data=df)
plt.title('BMI vs Stroke')
plt.show()
```

```python
df.groupby('stroke')['bmi'].mean()
```

```python
sns.boxplot(x='stroke', y='avg_glucose_level', data=df)
plt.title('Glucose Level vs Stroke')
plt.show()
```

```python
df.groupby('stroke')['avg_glucose_level'].mean()
```

```python
# Patients with stroke tend to have higher glucose levels compared to non-stroke patients
```

```python
pd.crosstab(df['smoking_status'], df['stroke'], normalize='index')
```
