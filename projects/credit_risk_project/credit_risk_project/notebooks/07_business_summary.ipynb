# Business Summary

This section summarizes the key business findings from the Credit Risk Analysis project.

The objective is to identify which borrower characteristics are associated with higher default risk and provide recommendations for lenders.

## Key Finding 1: Loan Grade Matters

Borrowers with lower loan grades (D-G) show significantly higher default rates than borrowers with grades A-C.

This suggests that loan grade is one of the strongest indicators of default risk.

## Key Finding 2: High Burden Borrowers Are Riskier

Borrowers in the high burden group have a much higher default rate compared with low burden borrowers.

This indicates that debt burden should be carefully evaluated during the loan approval process.

## Key Finding 3: Home Ownership Is Related to Default Risk

Renters show higher default rates than homeowners and borrowers with mortgages.

This may suggest lower financial stability among renter groups.

# Executive Summary

This project analyzed over 32,000 loan records to identify factors associated with borrower default risk.

Three major findings were identified:

1. Lower loan grades are associated with significantly higher default rates.
2. High debt burden borrowers are more likely to default.
3. Renters exhibit higher default risk than homeowners.

These findings can help lenders improve loan approval decisions and reduce financial losses.

# Business Recommendations

Based on the findings, lenders should:

- Apply additional review procedures for Grade D-G borrowers.
- Closely monitor borrowers with high debt burden.
- Consider housing status when evaluating credit risk.
- Use predictive models to identify high-risk borrowers early.
