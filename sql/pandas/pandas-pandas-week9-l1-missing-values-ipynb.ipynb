```python
# Week 9 — L1 Missing Values: constant / group-mean / conditional fills
import pandas as pd
import numpy as np

df = pd.DataFrame({'grp':['A','A','B','B'],'x':[1,None,3,None]})

# 1) constant fill
df_const = df.fillna({'x': 0})

# 2) group-mean fill
df['x_filled_mean'] = df.groupby('grp')['x'].transform(lambda s: s.fillna(s.mean()))

# 3) conditional fill: fill only missing values with 2
df['x_filled_cond'] = df['x'].where(df['x'].notna(), other=2)

# Save NA count artifact for portfolio
df.isna().sum().to_frame('na_count').to_csv('result_week9_na_counts.csv')
```
