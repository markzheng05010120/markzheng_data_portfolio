```python
import pandas as pd

# Sample long-format sales data
df = pd.DataFrame({
    'date': ['2024-01-01','2024-01-01','2024-01-02','2024-01-02','2024-01-03','2024-01-03'],
    'store': ['A','B','A','B','A','B'],
    'category': ['internet','internet','mobile','mobile','tv','tv'],
    'revenue': [120, 90, 60, 75, 40, 55]
})

df['date'] = pd.to_datetime(df['date'])

# 1) Pivot summary: total revenue by store x category
pv = pd.pivot_table(
    df,
    index='store',
    columns='category',
    values='revenue',
    aggfunc='sum',
    fill_value=0
).reset_index()

# 2) Melt back: wide -> long
long_again = pv.melt(id_vars=['store'], var_name='category', value_name='total_revenue')

# 3) Merge a mapping table
cat_map = pd.DataFrame({
    'category': ['internet','mobile','tv'],
    'biz_line': ['Connectivity','Wireless','Video']
})
final = long_again.merge(cat_map, on='category', how='left')

final.head()
```
