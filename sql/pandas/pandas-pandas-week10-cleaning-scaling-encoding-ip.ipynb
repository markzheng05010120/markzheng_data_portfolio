```python
# Week10 - scaling + encoding demo
import pandas as pd
from io import StringIO
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# ---- Part A: Character encoding demo ----
# Simulate reading a messy-encoding CSV (replace with your real file if you have one)
csv_bytes = 'name,city,price\nCafé,Montréal,12.5\nNiño,Bogotá,9.9\n'.encode('latin1', errors='replace')
df_enc = pd.read_csv(StringIO(csv_bytes.decode('latin1')), encoding='latin1')

# Save an artifact for portfolio
df_enc.to_csv('result_week10_encoding_demo.csv', index=False)

# ---- Part B: Scaling demo on numeric columns ----
data = pd.DataFrame({
    'product': ['A','B','C','D'],
    'qty': [3, 10, 5, 8],
    'price': [12.5, 5.0, 20.0, 7.5]
})

num_cols = ['qty','price']

# Standardization
std_scaler = StandardScaler()
data_std = data.copy()
data_std[num_cols] = std_scaler.fit_transform(data[num_cols])

# MinMax scaling
mm_scaler = MinMaxScaler()
data_mm = data.copy()
data_mm[num_cols] = mm_scaler.fit_transform(data[num_cols])

# Save artifacts
data_std.to_csv('result_week10_scaled_standard.csv', index=False)
data_mm.to_csv('result_week10_scaled_minmax.csv', index=False)
```
